Sel 1 — Import Library

```python
# Library manipulasi data
import pandas as pd
import numpy as np

# Library preprocessing
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# Library visualisasi
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')

print("Library berhasil diimport!")
```

Sel 2 — Upload Dataset

```python
from google.colab import files
import io

uploaded = files.upload()

# Mendapatkan nama file terbaru yang diunggah secara otomatis
file_name = list(uploaded.keys())[0]

# Membaca dataset sesuai dengan format file yang diunggah
try:
    df = pd.read_csv(io.BytesIO(uploaded[file_name]))
except Exception:
    df = pd.read_excel(io.BytesIO(uploaded[file_name]))

print(f"Dataset '{file_name}' berhasil dimuat!")
```

Sel 3 — Membuat Salinan Dataset

```python
# Membuat salinan dataset
df_preprocessing = df.copy()

# Melihat ukuran dataset
print("Jumlah baris:", df_preprocessing.shape[0])
print("Jumlah kolom:", df_preprocessing.shape[1])

display(df_preprocessing.head())
```

Sel 4 — Membersihkan Nama Kolom dan Format Data

```python
# Membersihkan nama kolom
df_preprocessing.columns = (
    df_preprocessing.columns.str.strip()
)

# Membersihkan spasi pada data kategorikal
categorical_columns = df_preprocessing.select_dtypes(
    include='object'
).columns

for column in categorical_columns:
    df_preprocessing[column] = (
        df_preprocessing[column].str.strip()
    )

print("Pembersihan format data berhasil!")

display(df_preprocessing.head())
```

Sel 5 — Menghapus Data Duplikat

```python
# Menghitung jumlah data sebelum penghapusan
jumlah_sebelum = len(df_preprocessing)

# Menghapus data duplikat
df_preprocessing = df_preprocessing.drop_duplicates()

# Menghitung jumlah data setelah penghapusan
jumlah_sesudah = len(df_preprocessing)

# Menghitung jumlah data yang dihapus
jumlah_dihapus = jumlah_sebelum - jumlah_sesudah

print("Jumlah data sebelum:", jumlah_sebelum)
print("Jumlah data duplikat yang dihapus:", jumlah_dihapus)
print("Jumlah data setelah:", jumlah_sesudah)
```

Sel 6 — Menangani Outlier

```python
# Kolom nilai akademik
score_columns = [
    'math score',
    'reading score',
    'writing score'
]

# Memastikan nilai berada dalam rentang 0-100
for column in score_columns:

    invalid_values = df_preprocessing[
        (df_preprocessing[column] < 0) |
        (df_preprocessing[column] > 100)
    ]

    print(f"\nKolom: {column}")
    print("Jumlah nilai tidak valid:", len(invalid_values))

print("\nNilai akademik dipertahankan selama masih valid.")
```

Sel 7 — Membuat Fitur Average Score

```python
# Menghitung rata-rata nilai siswa
df_preprocessing['average_score'] = (
    df_preprocessing['math score'] +
    df_preprocessing['reading score'] +
    df_preprocessing['writing score']
) / 3

# Menampilkan hasil
display(df_preprocessing.head())
```

Sel 8 — Melihat Hasil Feature Engineering

```python
# Menampilkan kolom nilai dan rata-rata
display(
    df_preprocessing[
        [
            'math score',
            'reading score',
            'writing score',
            'average_score'
        ]
    ].head(10)
)

# Statistik rata-rata nilai
print("\nStatistik Average Score:")
display(df_preprocessing['average_score'].describe())
```

Sel 9 — Menentukan Kolom Kategorikal

```python
# Mengidentifikasi kolom kategorikal
categorical_columns = df_preprocessing.select_dtypes(
    include='object'
).columns.tolist()

print("Kolom kategorikal:")
for column in categorical_columns:
    print("-", column)
```

Sel 10 — Melakukan One-Hot Encoding

```python
# Melakukan One-Hot Encoding
df_encoded = pd.get_dummies(
    df_preprocessing,
    columns=categorical_columns,
    dtype=int
)

# Menampilkan hasil encoding
display(df_encoded.head())

print("Jumlah kolom sebelum encoding:",
      df_preprocessing.shape[1])

print("Jumlah kolom setelah encoding:",
      df_encoded.shape[1])
```

Sel 11 — Standardisasi Menggunakan StandardScaler

```python
from sklearn.preprocessing import StandardScaler

# Menentukan kolom yang akan distandardisasi
numeric_columns = [
    'math score',
    'reading score',
    'writing score',
    'average_score'
]

# Membuat objek StandardScaler
scaler = StandardScaler()

# Membuat salinan dataset hasil encoding
df_scaled = df_encoded.copy()

# Melakukan standardisasi
df_scaled[numeric_columns] = scaler.fit_transform(
    df_scaled[numeric_columns]
)

print("Standardisasi berhasil dilakukan!")

display(df_scaled.head())
```

Sel 12 — Memeriksa Hasil Standardisasi

```python
# Melihat rata-rata setelah standardisasi
print("Rata-rata setelah standardisasi:")

display(
    df_scaled[numeric_columns].mean().round(2)
)

# Melihat standar deviasi
print("\nStandar deviasi setelah standardisasi:")

display(
    df_scaled[numeric_columns].std().round(2)
)
```

Sel 13 — Menyimpan Dataset

```python
# Menyimpan dataset hasil cleaning
df_preprocessing.to_csv(
    'StudentsPerformance_Cleaned.csv',
    index=False
)

# Menyimpan dataset hasil encoding dan scaling
df_scaled.to_csv(
    'StudentsPerformance_Preprocessed.csv',
    index=False
)

print("Dataset berhasil disimpan!")
```
